```python

# Step 1: Import required libraries

!pip install ucimlrepo -q

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

print("All libraries imported successfully!")
```

```python

# Step 2: Load the Ionosphere dataset

ionosphere = fetch_ucirepo(id=52)

# Separate input features and output labels
X = ionosphere.data.features
y = ionosphere.data.targets

# Convert target into a single column
y = y.iloc[:, 0]

print("Dataset loaded successfully!")
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python

# Step 3: Display first five records

df = X.copy()
df["Class"] = y.values

print(df.head())
```

```python

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nDataset Information:")
print(df.info())
```

```python

# Step 4: Check missing values

print("Missing values:")
print(df.isnull().sum().sum())

# Check duplicate records
print("\nDuplicate records:")
print(df.duplicated().sum())

# Check class distribution
print("\nClass distribution:")
print(y.value_counts())
```

```python

# Step 5: Class distribution graph

sns.countplot(x=y, hue=y, legend=False)

plt.title("Good vs Bad Radar Returns")
plt.xlabel("Radar Return Class")
plt.ylabel("Number of Records")

plt.show()
```

```python

# Step 5.2: Correlation heatmap

plt.figure(figsize=(14, 10))

sns.heatmap(
    X.corr(),
    cmap="coolwarm",
    center=0
)

plt.title("Radar Signal Feature Correlation")
plt.show()
```

```python

# Step 6: Convert class labels into numbers

y = y.map({
    "g": 1,
    "b": 0
})

print(y.head())
```

```python

# Step 7: Split dataset

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python

# Step 8: Import SVM

from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Create SVM model

svm_model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="rbf", C=1.0, gamma="scale"))
])

# Train the model

svm_model.fit(X_train, y_train)

print("SVM model trained successfully!")
```

```python

# Step 9: Make predictions

from sklearn.metrics import accuracy_score

y_pred = svm_model.predict(X_test)

# Calculate accuracy

accuracy = accuracy_score(y_test, y_pred)

print("SVM Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python

# Step 10: Model evaluation

from sklearn.metrics import classification_report

print("SVM Classification Report:\n")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Bad", "Good"]
    )
)
```

```python

# Step 11: Confusion Matrix

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
)

plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Bad", "Good"],
    yticklabels=["Bad", "Good"]
)

plt.title("SVM Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.show()
```

```python

# Step 12: Accuracy visualization

plt.figure(figsize=(5, 5))

plt.bar(
    ["SVM"],
    [accuracy * 100]
)

plt.title("SVM Model Accuracy")
plt.ylabel("Accuracy (%)")

plt.ylim(0, 100)

plt.show()
```

```python

# Step 13: Cross Validation

from sklearn.model_selection import cross_val_score, StratifiedKFold

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scores = cross_val_score(
    svm_model,
    X,
    y,
    cv=cv,
    scoring="accuracy"
)

print("Cross Validation Scores:", scores)

print(
    "Average Cross Validation Accuracy:",
    round(scores.mean() * 100, 2),
    "%"
)
```

```python

# Step 14: Save SVM model

import joblib

joblib.dump(
    svm_model,
    "radar_svm_model.pkl"
)

print("SVM model saved successfully!")
```

```python

# Step 15: Test prediction

sample = X_test.iloc[[0]]

prediction = svm_model.predict(sample)

if prediction[0] == 1:
    print("Radar Return: GOOD")
else:
    print("Radar Return: BAD")

print("Actual Class:", y_test.iloc[0])
```

```python

# Step 16: SVM Hyperparameter Tuning

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Create SVM pipeline
svm_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC())
])

# Parameters to test
param_grid = {
    "svm__C": [0.1, 1, 10, 100],
    "svm__gamma": ["scale", "auto", 0.01, 0.1],
    "svm__kernel": ["rbf", "linear"]
}

# Cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Search for best parameters
grid_search = GridSearchCV(
    svm_pipeline,
    param_grid,
    cv=cv,
    scoring="f1",
    n_jobs=-1
)

# Train and tune using training data only
grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV F1 Score:", round(grid_search.best_score_ * 100, 2), "%")

# Save the tuned model
best_svm_model = grid_search.best_estimator_
```

```python

# Step 17: Evaluate tuned SVM model

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

y_pred_tuned = best_svm_model.predict(X_test)

new_accuracy = accuracy_score(
    y_test,
    y_pred_tuned
)

print("Original SVM Accuracy: 95.77%")

print(
    "Tuned SVM Accuracy:",
    round(new_accuracy * 100, 2),
    "%"
)

print("\nClassification Report:\n")

print(
    classification_report(
        y_test,
        y_pred_tuned,
        target_names=["Bad", "Good"]
    )
)
```

```python

# Step 18: Tuned SVM confusion matrix

cm = confusion_matrix(
    y_test,
    y_pred_tuned,
    labels=[0, 1]
)

plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Bad", "Good"],
    yticklabels=["Bad", "Good"]
)

plt.title("Improved SVM Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.show()
```

```python

# Step 19: Save tuned SVM model

import joblib

joblib.dump(
    best_svm_model,
    "improved_radar_svm_model.pkl"
)

print("Improved SVM model saved successfully!")
```

```python

# Step 20.2: Prediction function

import gradio as gr
import numpy as np
import pandas as pd

def predict_radar(signal_input):

    try:
        # Convert entered values into numbers
        values = [
            float(x.strip())
            for x in signal_input.split(",")
        ]

        # Check number of features
        if len(values) != 34:
            return "Error: Please enter exactly 34 values."

        # Convert into DataFrame
        input_data = pd.DataFrame(
            [values],
            columns=X.columns
        )

        # Predict using trained SVM
        prediction = best_svm_model.predict(input_data)[0]

        if prediction == 1:
            return "GOOD RADAR RETURN"

        else:
            return "BAD RADAR RETURN"

    except ValueError:
        return "Error: Please enter only numerical values separated by commas."
```

```python

# Step 20.3: Create dashboard

sample_values = ", ".join(
    map(str, X_test.iloc[0].tolist())
)

with gr.Blocks(theme=gr.themes.Soft()) as dashboard:

    gr.Markdown(
        """
        # Radar Return Classification System

        ### AI/ML Based Ionospheric Signal Analysis

        Enter 34 radar signal attributes to classify
        the radar return as Good or Bad.
        """
    )

    gr.Markdown(
        """
        **Algorithm:** Support Vector Machine (SVM)

        **Kernel:** RBF

        **Test Accuracy:** {:.2f}%

        **Cross Validation Accuracy:** 94.01%
        """.format(new_accuracy * 100)
    )

    signal_input = gr.Textbox(
        label="Enter 34 Radar Signal Features",
        value=sample_values,
        lines=5,
        placeholder="Enter 34 comma-separated numerical values"
    )

    predict_button = gr.Button(
        "Predict Radar Return",
        variant="primary"
    )

    output = gr.Textbox(
        label="Classification Result"
    )

    predict_button.click(
        fn=predict_radar,
        inputs=signal_input,
        outputs=output
    )

    gr.Markdown(
        """
        **Project:** Radar Return Classification

        **Dataset:** UCI Ionosphere Dataset
        """
    )

dashboard.launch(share=True)
```
